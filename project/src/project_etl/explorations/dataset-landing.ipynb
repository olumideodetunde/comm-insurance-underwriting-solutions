# Landing Dataset of an actual motor vehicle insurance portfolio

- [Link to the dataset](https://data.mendeley.com/datasets/5cxyb5fp4f/2)

- Aim: To land the dataset to the `commercial-insurance-underwriting-solution-catalog` as raw as it is

```python
import os
import pandas as pd
from databricks.connect import DatabricksSession
spark = DatabricksSession.builder.getOrCreate()
```

## Create Catalog if it does not exist yet

```python
spark.sql("CREATE CATALOG IF NOT EXISTS `commercial-insurance-underwriting-solution-catalog`")
```

## Download the dataset
- Create a data path to download the data into
- Manually download the dataset from here - [Link to the dataset](https://data.mendeley.com/datasets/5cxyb5fp4f/2)
- Move the dataset to the folder

```python
os.makedirs('/Users/olumide/PycharmProjects/comm-insurance-underwriting-solutions/project/src/project_etl/data', exist_ok=True)
```

## Land Downloaded Dataset to Catalog
- Create the bronze schema
- Land the dataset and glossary with no transformation at all

```python
spark.sql("CREATE SCHEMA IF NOT EXISTS `commercial-insurance-underwriting-solution-catalog`.`bronze`")
```

### Claim dataset landing

```python
CSV_PATH = "/Users/olumide/PycharmProjects/comm-insurance-underwriting-solutions/project/src/project_etl/data/sample type claim.csv"
CATALOG = "`commercial-insurance-underwriting-solution-catalog`"
SCHEMA = "bronze"
TABLE = "claims"
```

```python
claims_df = pd.read_csv(CSV_PATH, delimiter=';')
claims_sdf = spark.createDataFrame(claims_df)
```

```python
claims_sdf.write.mode("overwrite").saveAsTable(f"{CATALOG}.{SCHEMA}.{TABLE}")
```
